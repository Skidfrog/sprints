## 4. Fes una petició POST per crear una nova publicació fictícia. Inclou el títol, el cos del missatge i un userId.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
peticio_post = {'userId': "BBB", 
                'title': 'Un post de prova', 
                'body': 'Aquest és un post de prova\nPertany al exercisi 4 del nivell 1\nIT Academy - Data Analitycs\nBarcelona Activa'}

response_post = requests.post(f"{url1}/posts", json = peticio_post)
try:
    response_post.raise_for_status()
except requests.exceptions.HTTPError as e:
        print(f"Error HTTP: {e}")

post = response_post.json()
print(post)

print("El estat de la petició es", response_post.status_code)
```

Llencem la petició per afegir noves dades amb post, i aprofitem el codi per al control d'errors. La importació és correcta i podem veure les dades en forma de diccionari. Observem que ha assignat una id 101 (el maxim que ja existia +1). El valor de codi de estat 201 confirma que la operació s'ha fet correctament.

## 6. Fes una petició DELETE sobre una publicació.

Mostra la resposta JSON.
Mostra el codi d'estat.

```python
response_delete = requests.delete(f"{url1}/posts/1")
try:
    response_delete.raise_for_status()
except requests.exceptions.HTTPError as e:
        print(f"Error HTTP: {e}")

delete = response_delete.json()
print(delete)

print("El estat de la petició es", response_delete.status_code)
```

## 1. Utilitza l’API d’Open Data BCN per:

Cercar un dataset del teu interès mitjançant package_search i package_show.

```python
bcn_od_url = "https://opendata-ajuntament.barcelona.cat/data/api/action/"
response_bcn = requests.get(f"{bcn_od_url}//package_list")
try:
    response_bcn.raise_for_status()
except requests.exceptions.HTTPError as e:
        print(f"Error HTTP: {e}")

bcn_data_list = response_bcn.json()

bcn_data_list["result"]
```

En primer terme, accedim al llistat de datasets actual a través del metode "/package_list", per a escollir un dataset.

## 2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.

```python
bcn_od_url = "https://opendata-ajuntament.barcelona.cat/data/api/action/"
response_agenda = requests.get(f"{bcn_od_url}/package_show?id=agenda-cultural")
try:
    response_agenda.raise_for_status()
except requests.exceptions.HTTPError as e:
        print(f"Error HTTP: {e}")

agenda = response_agenda.json()

agenda
```

Ente el llistat, trobem agenda-cultural, que disposa de recursos en format .csv. Busquem el resource id que mostra el json del recurs:

"id': '3abb2414-1ee0-446e-9c25-380e938adb73"

## 3. Amb el resource_id del recurs seleccionat, realitza una consulta mitjançant datastore_search per:

Recuperar almenys 100 registres.

```python
resource_id = "3abb2414-1ee0-446e-9c25-380e938adb73"

url_search = f"{bcn_od_url}/datastore_search?resource_id={resource_id}&limit=200"

response_search = requests.get(url_search)
response_search.raise_for_status()

data_search = response_search.json()
data_search["result"]["records"]
```

Configurem la cerca amb l'ID del dataset i hi posem un limit de 200, en observar que el resultat conté molts camps buits, per tal de poder tenir una major visió del tipus de dades que trobarem en aquest dataset. 

Desprès en filtrem els resultats per a confeccionar el DataFrame i posteriorment, exportar com a CSV.

## 4. Converteix els resultats a un DataFrame.

```python
df = pd.DataFrame(data_search["result"]["records"])
df
```
